```python
# ==============================================================================
# FEDSCREEN PIPELINE — STEP 1
# DATASET AUDIT & PATIENT-LEVEL SPLITTING
# ==============================================================================
#
# Target Paths (Updated to Google Drive directly):
#   - /content/drive/MyDrive/fedscreen/dataset/Malaria
#   - /content/drive/MyDrive/fedscreen/dataset/all-idb-1
#   - /content/drive/MyDrive/fedscreen/dataset/all-idb-2
#   - /content/drive/MyDrive/fedscreen/dataset/AneRBC
#
# Purpose:
#   1. Verify the Colab environment and GPU
#   2. Scan all datasets and collect image metadata
#   3. Extract patient/slide-level group IDs
#   4. Create 70% Train / 15% Validation / 15% Test splits
#   5. Prevent patient/slide-level data leakage
#   6. Save split metadata for the next pipeline step
# ==============================================================================


import os
import re

import numpy as np
import pandas as pd
import torch
from sklearn.model_selection import GroupShuffleSplit


# ==============================================================================
# 0. ENVIRONMENT & HARDWARE VERIFICATION
# ==============================================================================

print("=" * 60)
print("1. VERIFYING COLAB ENVIRONMENT")
print("=" * 60)

print(f"PyTorch Version : {torch.__version__}")

gpu_available = torch.cuda.is_available()
print(f"CUDA Available  : {gpu_available}")

if gpu_available:
    print(f"Device Name     : {torch.cuda.get_device_name(0)}")
else:
    print("⚠️ Running on CPU.")
    print("   Enable GPU in Runtime → Change runtime type.")


# Set seed for reproducible splits
SEED = 42

np.random.seed(SEED)
torch.manual_seed(SEED)


# ==============================================================================
# 1. DATASET PATH DEFINITIONS & ID EXTRACTION RULES
# ==============================================================================

DATASET_PATHS = {
    "Malaria": "/content/drive/MyDrive/fedscreen/dataset/Malaria",
    "ALL-IDB-1": "/content/drive/MyDrive/fedscreen/dataset/all-idb-1",
    "ALL-IDB-2": "/content/drive/MyDrive/fedscreen/dataset/all-idb-2",
    "AneRBC": "/content/drive/MyDrive/fedscreen/dataset/AneRBC",
}


def extract_group_id(
    dataset_key: str,
    filename: str,
    relative_path: str
) -> str:
    """
    Extract Patient or Slide IDs to prevent data leakage during splitting.
    """

    # --------------------------------------------------------------------------
    # Malaria Dataset
    # Example:
    # C33P1thinF_IMG_20150619_114756a_cell_179.png
    #                         ↓
    #                    Patient_1
    # --------------------------------------------------------------------------

    if dataset_key == "Malaria":

        p_match = re.search(r"P(\d+)", filename)

        if p_match:
            return f"Malaria_Patient_{p_match.group(1)}"

        s_match = re.search(r"C(\d+)", filename)

        if s_match:
            return f"Malaria_Slide_{s_match.group(1)}"

        return f"Malaria_Group_{filename[:6]}"


    # --------------------------------------------------------------------------
    # ALL-IDB-1 / ALL-IDB-2
    # Example:
    # Im001_1.jpg
    #      ↓
    #   Slide_001
    # --------------------------------------------------------------------------

    elif dataset_key in ["ALL-IDB-1", "ALL-IDB-2"]:

        match = re.search(r"Im(\d+)", filename, re.IGNORECASE)

        if match:
            return f"ALL_IDB_Slide_{match.group(1)}"

        return f"ALL_IDB_Group_{filename[:5]}"


    # --------------------------------------------------------------------------
    # AneRBC Dataset
    # Group by parent subfolder or sample prefix
    # --------------------------------------------------------------------------

    elif dataset_key == "AneRBC":

        parent_folder = os.path.basename(
            os.path.dirname(relative_path)
        )

        prefix = (
            filename.split("_")[0]
            if "_" in filename
            else filename[:5]
        )

        return f"AneRBC_{parent_folder}_{prefix}"


    # --------------------------------------------------------------------------
    # Fallback
    # --------------------------------------------------------------------------

    return f"Group_{filename}"


# ==============================================================================
# 2. SCANNING DATASET DIRECTORIES IN GOOGLE DRIVE
# ==============================================================================

print("\n" + "=" * 60)
print("2. SCANNING DATASET DIRECTORIES")
print("=" * 60)


records = []

valid_extensions = (
    ".png",
    ".jpg",
    ".jpeg",
    ".tif",
    ".tiff",
    ".bmp",
)


for key, path in DATASET_PATHS.items():

    # Check whether dataset directory exists
    if not os.path.exists(path):

        print(f"⚠️ Warning: Path does not exist: {path}")
        continue

    file_count = 0

    # Recursively scan all files
    for root, dirs, files in os.walk(path):

        for fname in files:

            # Process only supported image formats
            if fname.lower().endswith(valid_extensions):

                full_path = os.path.join(root, fname)

                # Path relative to dataset root
                rel_path = os.path.relpath(full_path, path)

                # --------------------------------------------------------------
                # Derive class/category from folder structure
                # --------------------------------------------------------------

                subfolders = rel_path.split(os.sep)

                category = (
                    subfolders[0]
                    if len(subfolders) > 1
                    else "General"
                )

                # --------------------------------------------------------------
                # Extract patient/slide-level group ID
                # --------------------------------------------------------------

                group_id = extract_group_id(
                    key,
                    fname,
                    full_path
                )

                # --------------------------------------------------------------
                # Store metadata
                # --------------------------------------------------------------

                records.append({
                    "dataset": key,
                    "filename": fname,
                    "file_path": full_path,
                    "category": category,
                    "group_id": group_id,
                })

                file_count += 1

    print(
        f"Found {file_count:>6,} images "
        f"in '{key}' ({path})"
    )


# ==============================================================================
# CREATE MASTER METADATA DATAFRAME
# ==============================================================================

df_meta = pd.DataFrame(records)


if len(df_meta) == 0:

    raise FileNotFoundError(
        "No images were found in the specified paths! "
        "Please check if your dataset folders are unzipped under your Google Drive paths."
    )


print("-" * 60)

print(
    f"Total Combined Dataset Images : "
    f"{len(df_meta):,}"
)

print(
    f"Total Unique Group/Patient IDs : "
    f"{df_meta['group_id'].nunique():,}"
)


# ==============================================================================
# 3. EXECUTING GROUP-BASED SPLITTING
#    70% TRAIN / 15% VALIDATION / 15% TEST
# ==============================================================================

print("\n" + "=" * 60)
print("3. EXECUTING PATIENT/SLIDE-LEVEL GROUP SPLITTING")
print("=" * 60)


# ------------------------------------------------------------------------------
# Step A: Split off 15% Test Set
# ------------------------------------------------------------------------------

gss_test = GroupShuffleSplit(
    n_splits=1,
    test_size=0.15,
    random_state=SEED
)

train_val_idx, test_idx = next(
    gss_test.split(
        df_meta,
        groups=df_meta["group_id"]
    )
)


df_train_val = (
    df_meta
    .iloc[train_val_idx]
    .copy()
    .reset_index(drop=True)
)

df_test = (
    df_meta
    .iloc[test_idx]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------------------------
# Step B: Split remaining 85%
#          → Train = 70%
#          → Validation = 15%
#
# 15 / 85 = 0.1765
# ------------------------------------------------------------------------------

gss_val = GroupShuffleSplit(
    n_splits=1,
    test_size=0.1765,
    random_state=SEED
)

train_idx, val_idx = next(
    gss_val.split(
        df_train_val,
        groups=df_train_val["group_id"]
    )
)


df_train = (
    df_train_val
    .iloc[train_idx]
    .copy()
    .reset_index(drop=True)
)

df_val = (
    df_train_val
    .iloc[val_idx]
    .copy()
    .reset_index(drop=True)
)


# ------------------------------------------------------------------------------
# Assign split labels
# ------------------------------------------------------------------------------

df_train["split"] = "train"
df_val["split"] = "val"
df_test["split"] = "test"


# ------------------------------------------------------------------------------
# Combine all splits
# ------------------------------------------------------------------------------

df_full_split = pd.concat(
    [
        df_train,
        df_val,
        df_test
    ],
    axis=0
).reset_index(drop=True)


# ==============================================================================
# 4. RIGOROUS DATA LEAKAGE CHECK
# ==============================================================================

print("\n4. VERIFYING ZERO DATA LEAKAGE")
print("-" * 60)


# Get unique patient/slide groups for each split
train_groups = set(df_train["group_id"])
val_groups = set(df_val["group_id"])
test_groups = set(df_test["group_id"])


# Check for overlapping groups
overlap_tv = train_groups.intersection(val_groups)
overlap_tt = train_groups.intersection(test_groups)
overlap_vt = val_groups.intersection(test_groups)


# ------------------------------------------------------------------------------
# Assertions: There must be ZERO overlap
# ------------------------------------------------------------------------------

assert len(overlap_tv) == 0, (
    f"Leakage between Train & Val: {overlap_tv}"
)

assert len(overlap_tt) == 0, (
    f"Leakage between Train & Test: {overlap_tt}"
)

assert len(overlap_vt) == 0, (
    f"Leakage between Val & Test: {overlap_vt}"
)


print(
    "✅ PASSED: 0% Data Leakage! "
    "No patient/slide overlaps across Train, Val, and Test."
)


# ==============================================================================
# SPLIT BREAKDOWN TABLE
# ==============================================================================

print("\nOverall Split Summary:")


summary = (
    df_full_split
    .groupby(["split", "dataset"])
    .agg(
        total_images=("filename", "count"),
        unique_groups=("group_id", "nunique")
    )
    .reset_index()
)


print(summary.to_string(index=False))


# ==============================================================================
# 5. SAVE SPLIT INDEX CSV
# ==============================================================================

output_dir = "/content/drive/MyDrive/fedscreen/results"
os.makedirs(output_dir, exist_ok=True)
output_csv = os.path.join(output_dir, "fedscreen_split_metadata.csv")

df_full_split.to_csv(
    output_csv,
    index=False
)


print("\n" + "=" * 60)
print(f"💾 Saved metadata file: '{output_csv}'")
print(
    "This CSV will be used directly in Step 2 "
    "for cell segmentation & preprocessing."
)
print("=" * 60)
```

```python
import os
import sys
import subprocess

# Ensure OpenCV is installed
try:
    import cv2
except ImportError:
    print("Installing opencv-python-headless...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "opencv-python-headless"])
    import cv2

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

print("=" * 60)
print("1. CONNECTING TO PRE-EXISTING PREPROCESSED DATA")
print("=" * 60)

# Target input/output paths
PREPROCESSED_DIR = "/content/drive/MyDrive/fedscreen/preprocessed"
step1_csv = "/content/drive/MyDrive/fedscreen/results/fedscreen_split_metadata.csv"
backup_meta_csv = "/content/drive/MyDrive/fedscreen/metadata_backups/fedscreen_preprocessed_metadata.csv"
output_dir = "/content/drive/MyDrive/fedscreen/results"
os.makedirs(output_dir, exist_ok=True)
output_meta_csv = os.path.join(output_dir, "fedscreen_preprocessed_metadata.csv")

# Fallback 1: If a complete preprocessed metadata backup exists, restore it directly to bypass drive lister timeouts
if os.path.exists(backup_meta_csv):
    print(f"✅ Found existing preprocessed metadata backup at: {backup_meta_csv}")
    df_preprocessed = pd.read_csv(backup_meta_csv)
    df_preprocessed.to_csv(output_meta_csv, index=False)
    print(f"✅ Restored and verified preprocessed metadata connection: {output_meta_csv}")
elif os.path.exists(step1_csv):
    df_meta = pd.read_csv(step1_csv)
    print(f"✅ Successfully loaded input split metadata from: {step1_csv}")
    processed_records = []
    print(f"Verifying pre-existing preprocessed crops folder at: {PREPROCESSED_DIR}")

    # Retrieve available filenames while handling potential I/O errors gracefully
    available_files = {}
    if os.path.exists(PREPROCESSED_DIR):
        try:
            available_files = {f.lower(): f for f in os.listdir(PREPROCESSED_DIR)}
            print(f"ጃ Found {len(available_files):,} preprocessed files in directory.")
        except OSError as e:
            print(f"☑ Warning: Could not list directory due to GDrive I/O limitations: {e}")
            print("Reverting to direct mapping verification...")
    else:
        print("☑ Preprocessed directory not found on Drive!")

    for idx, row in df_meta.iterrows():
        save_fname = f"proc_{row['dataset']}_{idx}_{row['filename']}"
        if not save_fname.lower().endswith(".png"):
            save_fname += ".png"

        save_path = os.path.join(PREPROCESSED_DIR, save_fname)
        matched = False

        if len(available_files) > 0:
            if save_fname.lower() in available_files:
                save_path = os.path.join(PREPROCESSED_DIR, available_files[save_fname.lower()])
                matched = True
            else:
                alt_name = f"{row['filename']}"
                if not alt_name.lower().endswith(".png"):
                    alt_name += ".png"
                for key_lower, real_name in available_files.items():
                    if alt_name.lower() in key_lower:
                        save_path = os.path.join(PREPROCESSED_DIR, real_name)
                        matched = True
                        break
        else:
            # Direct file check if listdir failed due to Drive I/O timeout
            if os.path.exists(save_path):
                matched = True

        if matched:
            row_dict = row.to_dict()
            row_dict["preprocessed_path"] = save_path
            processed_records.append(row_dict)

    df_preprocessed = pd.DataFrame(processed_records)
    df_preprocessed.to_csv(output_meta_csv, index=False)
    print(f"✅ Connected metadata saved to: '{output_meta_csv}'")
    print(f"✅ Verified and compiled metadata for {len(df_preprocessed):,} crops.")
else:
    raise FileNotFoundError(f"Could not find input metadata splits or backups. Please verify your Google Drive paths.")

# Quick validation plot
print("\n" + "=" * 60)
print("2. VISUAL QUALITY CHECK FROM DRIVE")
print("=" * 60)

if len(df_preprocessed) > 0:
    # Only plot if we can find files that exist locally
    valid_sample_paths = []
    for _, row in df_preprocessed.iterrows():
        if os.path.exists(row["preprocessed_path"]):
            valid_sample_paths.append(row)
        if len(valid_sample_paths) >= 4:
            break

    if len(valid_sample_paths) > 0:
        fig, axes = plt.subplots(1, len(valid_sample_paths), figsize=(16, 4))
        if len(valid_sample_paths) == 1:
            axes = [axes]
        for i, row in enumerate(valid_sample_paths):
            proc_rgb = np.array(Image.open(row["preprocessed_path"]))
            axes[i].imshow(proc_rgb)
            axes[i].set_title(f"{row['dataset']} Crop\n({row['category'][:30]}...)", fontsize=10)
            axes[i].axis("off")
        plt.tight_layout()
        plt.show()
    else:
        print("☑ Found metadata but files cannot be read directly at this moment due to Google Drive limitations.")
else:
    print("★ No preprocessed files mapped! Please check the folder paths and filenames.")

print("\n" + "=" * 60)
print("★ STEP 2 BYPASS COMPLETED SUCCESSFULLY!")
print(f"ጃ Connected metadata saved to: '{output_meta_csv}'")
print("=" * 60)
```

```python
# ==============================================================================
# FEDSCREEN PIPELINE — STEP 3 BYPASS
# CONNECTING PRE-GENERATED NORMALIZED IMAGES
# ==============================================================================
#
# Inputs:
#   - /content/drive/MyDrive/fedscreen/results/fedscreen_preprocessed_metadata.csv
#   - Pre-existing normalized crops in /content/drive/MyDrive/fedscreen/normalized
#
# Outputs:
#   - Verified normalized metadata CSV stored in results subfolder of GDrive
#
# ==============================================================================

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from PIL import Image

print("=" * 60)
print("1. LOADING STEP 2 PREPROCESSED METADATA")
print("=" * 60)

# Source path specified by the user
input_csv = "/content/drive/MyDrive/fedscreen/results/fedscreen_preprocessed_metadata.csv"

if not os.path.exists(input_csv):
    raise FileNotFoundError(f"Could not find Step 2 preprocessed metadata at: {input_csv}")

df_meta = pd.read_csv(input_csv)
print(f"✅ Loaded {len(df_meta):,} preprocessed image records.")

# ==============================================================================
# DEFINE INPUT/OUTPUT DIRECTORIES
# ==============================================================================
NORMALIZED_DIR = "/content/drive/MyDrive/fedscreen/normalized"
output_meta_csv = "/content/drive/MyDrive/fedscreen/results/fedscreen_normalized_metadata.csv"

print(f"\n᎐ Verifying pre-existing normalized crops folder at: {NORMALIZED_DIR}")
if not os.path.exists(NORMALIZED_DIR):
    raise FileNotFoundError(f"Normalized crops directory not found at: {NORMALIZED_DIR}")

# Retrieve available filenames to handle I/O robustly
available_files = {}
try:
    available_files = {f.lower(): f for f in os.listdir(NORMALIZED_DIR)}
    print(f"✅ Found {len(available_files):,} files in normalized directory.")
except OSError as e:
    print(f"☑ Warning: Could not list directory directly due to GDrive limitations: {e}")
    print("Reverting to direct mapping verification...")

# ==============================================================================
# 2. MATCH & VERIFY METADATA WITH DRIVE FILENAMES (BYPASS ROTATION)
# ==============================================================================
print("\n" + "=" * 60)
print("2. EXECUTING METADATA MAPPING")
print("=" * 60)

norm_records = []

for idx, row in df_meta.iterrows():
    ds_key = row["dataset"]
    fname = row["filename"]

    # Reconstruct expected normalized filename template
    save_fname = f"norm_{ds_key}_{idx}_{fname}"
    if not save_fname.lower().endswith(".png"):
        save_fname += ".png"

    save_path = os.path.join(NORMALIZED_DIR, save_fname)
    matched = False

    if len(available_files) > 0:
        if save_fname.lower() in available_files:
            save_path = os.path.join(NORMALIZED_DIR, available_files[save_fname.lower()])
            matched = True
        else:
            # Try absolute raw filename search fallback
            alt_name = f"{fname}"
            if not alt_name.lower().endswith(".png"):
                alt_name += ".png"
            for key_lower, real_name in available_files.items():
                if alt_name.lower() in key_lower:
                    save_path = os.path.join(NORMALIZED_DIR, real_name)
                    matched = True
                    break
    else:
        # Fallback to direct path check if listdir timed out
        if os.path.exists(save_path):
            matched = True

    if matched:
        row_dict = row.to_dict()
        row_dict["normalized_path"] = save_path
        norm_records.append(row_dict)

df_normalized = pd.DataFrame(norm_records)

# Save normalized metadata to the results subfolder
os.makedirs(os.path.dirname(output_meta_csv), exist_ok=True)
df_normalized.to_csv(output_meta_csv, index=False)

print(f"\n✅ Successfully verified and mapped {len(df_normalized):,} normalized cell images!")
print(f"᎐ Updated Metadata Saved To: '{output_meta_csv}'")

# ==============================================================================
# 3. VISUAL QUALITY CHECK COMPARISON
# ==============================================================================
print("\n" + "=" * 60)
print("3. VISUAL QUALITY CHECK COMPARISON")
print("=" * 60)

if len(df_normalized) > 0:
    # Only plot if we can find files that exist on GDrive
    valid_samples = []
    for _, row in df_normalized.iterrows():
        if os.path.exists(row["normalized_path"]):
            valid_samples.append(row)
        if len(valid_samples) >= 4:
            break

    if len(valid_samples) > 0:
        fig, axes = plt.subplots(2, len(valid_samples), figsize=(16, 8))
        fig.suptitle("FedScreen Step 3: Preprocessed vs Normalization Bypass Check", fontsize=14, fontweight="bold")

        for i, row in enumerate(valid_samples):
            proc_path = row["preprocessed_path"]
            norm_path = row["normalized_path"]

            before_rgb = np.array(Image.open(proc_path).convert("RGB"))
            after_rgb = np.array(Image.open(norm_path).convert("RGB"))

            # Before
            axes[0, i].imshow(before_rgb)
            axes[0, i].set_title(f"Preprocessed: {row['dataset']}\n({row['category'][:20]}...)", fontsize=9)
            axes[0, i].axis("off")

            # After
            axes[1, i].imshow(after_rgb)
            axes[1, i].set_title("Macenko Normalized\n(Pre-existing)", fontsize=9)
            axes[1, i].axis("off")

        plt.tight_layout()
        plt.show()
    else:
        print("☑ Found matched metadata records but physical image files are temporarily unreadable on GDrive.")
else:
    print("☕ No normalized files could be mapped. Please verify your normalized folder contents.")

print("\n" + "=" * 60)
print("᎐ STEP 3 BYPASS COMPLETED SUCCESSFULLY!")
print("=" * 60)
```
